```python
from pydantic import BaseModel,Field
from typing import List

class Quote(BaseModel):
    """Motivational quotes details"""
    quote:str=Field(...,description="Motivational quotes")
    hashtag:List[str]=Field(...,description="List of 6 hashtag for facebook related to quotes")

class QuoteList(BaseModel):
    """ Motivational , self help , Quotes list """
    quotes: List[Quote]  
```

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['OPENAI_API_KEY']=os.getenv("OPENAI_API_KEY")
```

```python
### ChatOpenAI

from langchain_openai import ChatOpenAI
model=ChatOpenAI(model="gpt-4.1")
model_with_output = model.with_structured_output(QuoteList)
```

```python
prompt = "You are motivation and self help content writer. You have to give me 5 motivation quotes which" \
"encourage and boost morales."
response=model_with_output.invoke(prompt)
```

```python
def splittext(sentence,wordCount):
    words = sentence.split()

    # Group words into chunks of 3 using list comprehension and slicing
    word_groups = [words[i:i+wordCount] for i in range(0, len(words), wordCount)]

    # Join each group back into a string
    result_strings = [' '.join(group) for group in word_groups]

    print(result_strings)
    return result_strings
```

```python
COUNTER_FILE = '/mnt/c/edit/images/counter.txt'

def update_counter():
    """Reads the current counter value from a file, increments it, and writes the new value back."""
    with open(COUNTER_FILE, 'r') as f:
        count = int(f.read())
    count += 1
    with open(COUNTER_FILE, 'w') as f:
        f.write(str(count))
    return count
```

```python
from PIL import Image, ImageDraw, ImageFont

def text_to_image(quote:Quote):
    # ---------- CONFIG ----------
    WIDTH, HEIGHT = 1080, 1920   # Story size
    BG_COLOR = (10, 10, 10)      # Dark grey background
    TEXT_COLOR = (255, 255, 255)

    TEXT = quote.quote
    TEXT = splittext(TEXT,3)
    SENT = "\n\n ".join(TEXT)

    # FONT_PATH = "seguiemj.ttf"   # Change if needed
    FONT_PATH = "Roboto.ttf"   # Change if needed
    FONT_SIZE = 100
    LINE_SPACING = 18
    # ----------------------------

    # Create image
    img = Image.new("RGB", (WIDTH, HEIGHT), BG_COLOR)
    draw = ImageDraw.Draw(img)

    # Load font
    font = ImageFont.truetype(FONT_PATH, FONT_SIZE)

    # Calculate text size
    bbox = draw.multiline_textbbox((0, 0), SENT, font=font, spacing=LINE_SPACING)
    text_width = bbox[2] - bbox[0]
    text_height = bbox[3] - bbox[1]

    # Center position
    x = (WIDTH - text_width) // 2
    y = (HEIGHT - text_height) // 2

    # Draw text
    draw.multiline_text(
        (x, y),
        SENT,
        fill=TEXT_COLOR,
        font=font,
        align="center",
        spacing=LINE_SPACING
    )
    file_no = update_counter();
    # Save image
    file_name = f"/mnt/c/edit/images/quotes_{file_no}.png" 
    img.save(file_name, quality=95)
    print("Image generated: ",file_name)
```

```python
quotes = response
for q in quotes.quotes:
    text_to_image(q)
```

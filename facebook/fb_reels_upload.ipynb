## Upload stage

```python

url = response.json()['upload_url']

VIDEO_FILE_PATH="reels.mp4"
video_len = os.path.getsize(VIDEO_FILE_PATH)

headers = {
	"Authorization": f"OAuth {ACCESS_TOKEN}",
	"offset": '0',
    "file_size":f"{video_len}",
    "Content-Type":"video/mp4"
}
files = {'file': open(VIDEO_FILE_PATH, 'rb')}
response_upload = requests.post(url,headers=headers, files=files)
response_upload
```
